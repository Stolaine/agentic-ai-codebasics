```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model = 'gemma-4-31b-it')

response = llm.invoke("How many moons does Jupiter have?")
```

```python
print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are a helpful assistant that answers in short one line"],
    ["human", "How many moons does Jupiter have?"]
])
```

```python
print(response.text)
```

```python
llm_with_temp_0 = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)
llm_with_temp_1 = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=1)
```

```python
prompt = "Write a single tagline for a qirky coffee shop. Only output the tagline, nothing else."
```

```python
response_with_temp_0 = llm_with_temp_0.invoke(prompt)
response_with_temp_1 = llm_with_temp_1.invoke(prompt)
```

```python
print(response_with_temp_0.text)
print(response_with_temp_1.text)
```

```python
from langchain_groq import ChatGroq
```

```python
llm_from_groq = ChatGroq(model="qwen/qwen3-32b", reasoning_format="parsed")
```

```python
messages = [
    ("system", "You are a helpful assistant that gives a short answer to a question."),
    ("human", "How many moons does Jupiter have?")
]
```

```python
response_from_groq = llm_from_groq.invoke(messages)
```

```python
print(response_from_groq.text)
```
